```python
import matplotlib.pyplot as plt

hours = [1,2,3,4,5,6,7,8]
visitors = [50,60,65,70,90,120,110,130]

plt.plot(hours,visitors,
         color='green',
         marker='s',
         linestyle='--',
         linewidth=2,
         alpha=1,
         label='Visitors')

plt.title('Website Visitors per Hour')
plt.xlabel('Hours of day')
plt.ylabel('Numnbers of visitors')
plt.legend()

plt.grid(True,linestyle='-',alpha=1)
plt.show()
```

```python
import matplotlib.pyplot as plt

products = ['Laptop','Mobile','Tablet','Headphones','Smartwatch']
sales = [250,400,150,300,180]

plt.bar(products,sales,color='green',edgecolor='red',alpha=1,width=0.5)
plt.title('Product Sales Comparision')
plt.xlabel('Products')
plt.ylabel('Units Sold')
plt.show()
```

```python
import matplotlib.pyplot as plt

brands = ['Apple','Samsung','Xiaomi','Oppo','Others']
market_share = [30,25,20,10,15]
plt.pie(market_share,
        labels=brands,
        autopct='%.2f%%',
        startangle=90,
        colors=['gold','skyblue','lightcoral','lightgreen','violet'],
        explode=(0.2,0,0,0,0),
        shadow=True)

plt.title('Mobile Market Share 2025')
plt.show()
```

```python
import matplotlib.pyplot as plt

ad_spend = [2,4,5,7,8,10,11,12]
sales = [27,25,28,35,40,48,50,52]
profit = [200,250,90,350,40,80,500,520]
colors = [8,1,2,3,5,6,7,4]

plt.scatter(ad_spend,sales,
            s=profit,
            c=colors,
            cmap='Reds',
            alpha=0.8,
            edgecolors='black',
            linewidth=1.2
            )

plt.title('Advertising Spend vs Sales (with color & size)',fontsize=14,fontweight='bold')
plt.xlabel('Ad Spend ($1000s)',fontsize=12)
plt.ylabel('Sales ($1000s)',fontsize=12)

plt.colorbar(label='Sales Values')
plt.grid(True,linestyle='--',alpha=0.2)
plt.show()
```

```python
import matplotlib.pyplot as plt

scores = [
    [70,67,89,67,89,67,56,78],
    [78,43,67,67,89,78,85,68],
    [68,46,89,86,56,78,89,56],
    [78,45,34,87,76,98,56,89],
    [67,34,56,97,56,68,68,55]
]

plt.boxplot(scores,
            patch_artist=True,
            medianprops=dict(color='red',linewidth=1),
            whiskerprops=dict(color='green',linewidth=1),
            capprops=dict(color='black',linewidth=1),
            notch=True,
            vert=True,
            labels=['A','B','C','D','E']
            )

plt.title('Exam Scores Distribution Across Classes')
plt.xlabel('Classes')
plt.ylabel('Scores')
plt.grid(True,linestyle='--',alpha=0.7)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

data = np.array([
   [30,45,23,45,24,46,35],
   [57,23,45,24,46,32,24],
   [56,45,67,34,56,45,34],
   [53,23,45,56,23,23,45],
   [34,23,34,56,34,23,45]
])

days = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
week = ['Week 1','Week 2','Week 3','Week 4','Week 5']

sns.heatmap(data,
            annot=True,
            fmt='.2f',
            cmap='YlOrRd',
            linewidth=0.5,
            xticklabels=days,
            yticklabels=week)

plt.title('Weekly Temperature heatmap',fontsize=14,fontweight='bold')
plt.xlabel('Days of the week')
plt.ylabel('Weeks')
plt.show()
```
